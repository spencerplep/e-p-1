```python
%gui qt

from e_p_1.model import Object, Space


import numpy as np
import pyqtgraph as qt
view = qt.ImageView()
```

```python

a = Object(size=(3, 10, 1), position=(0, 0, 6))

b = Object(size=(3, 10, 1), position=(0, 0, -6))

space = Space(space_limits=[[-5, -5, -5], [5, 5, 5]], objects=[a, b])

grids = space.build_grids()

grids = space.ambient_field(grids, direction=[0, 0, 1], magnitude=-100.0)

v = grids.voltage[0, :, :]
```

```python


view.setImage(v)
view.show()

view.ui.histogram.hide()

view.ui.roiPlot.hide()
view.ui.normGroup.hide()
```
